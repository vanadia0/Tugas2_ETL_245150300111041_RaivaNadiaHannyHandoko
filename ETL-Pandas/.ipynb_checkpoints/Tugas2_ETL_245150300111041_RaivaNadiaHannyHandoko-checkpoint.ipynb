```python
from pathlib import Path
import json
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

warnings.filterwarnings('ignore')
pd.set_option('display.max_columns', 40)
pd.set_option('display.width', 140)
sns.set_theme(style='whitegrid', palette='colorblind')

project_candidates = [Path.cwd(), *Path.cwd().parents]
PROJECT_ROOT = next((path for path in project_candidates if (path / 'data' / 'raw').is_dir()), Path.cwd())
RAW_DIR = PROJECT_ROOT / 'data' / 'raw'
OUTPUT_DIR = PROJECT_ROOT / 'data' / 'output'
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

EXPECTED_COLUMNS = ['timestamp', 'sensor_id', 'location', 'temperature_celsius', 'humidity_percent', 'pressure_hpa', 'air_quality_aqi', 'status']
MEASURE_COLUMNS = ['temperature_celsius', 'humidity_percent', 'pressure_hpa', 'air_quality_aqi']
VALID_RANGES = {'temperature_celsius': (-10, 50), 'humidity_percent': (0, 100), 'pressure_hpa': (950, 1050), 'air_quality_aqi': (0, 500)}
print('Folder proyek:', PROJECT_ROOT)
print('Pandas:', pd.__version__)
```

## 1. Extract

Setiap sumber dibaca dengan penanganan `FileNotFoundError`. Nama kolom distandarkan sejak ekstraksi agar seluruh sumber dapat disusun ke skema pengamatan yang seragam. Kolom `_source` menyimpan asal baris untuk penelusuran.

```python
def read_optional_csv(file_path, separator=','):
    try:
        frame = pd.read_csv(file_path, sep=separator)
        print('Terbaca:', file_path.name, frame.shape)
        return frame
    except FileNotFoundError:
        print('File tidak ditemukan, dilewati:', file_path)
        return None

source_frames = []
csv_sources = [
    (RAW_DIR / 'sensor_data_main.csv', ',', 'sensor_data_main.csv'),
    (RAW_DIR / 'sensor_data_semicolon.csv', ';', 'sensor_data_semicolon.csv'),
]
for file_path, separator, label in csv_sources:
    frame = read_optional_csv(file_path, separator)
    if frame is not None:
        frame['_source'] = label
        source_frames.append(frame)

excel_path = RAW_DIR / 'sensor_data_monthly.xlsx'
try:
    excel_sheets = pd.read_excel(excel_path, sheet_name=None)
    print('Sheet Excel:', list(excel_sheets))
    for sheet_name, frame in excel_sheets.items():
        frame['_source'] = 'sensor_data_monthly.xlsx:' + sheet_name
        source_frames.append(frame)
except FileNotFoundError:
    print('File tidak ditemukan, dilewati:', excel_path)

config_path = RAW_DIR / 'sensor_config.json'
try:
    with config_path.open(encoding='utf-8') as config_file:
        sensor_config = json.load(config_file)
    print('Konfigurasi sensor terbaca:', len(sensor_config))
except FileNotFoundError:
    print('File tidak ditemukan, konfigurasi kosong:', config_path)
    sensor_config = []

aligned_frames = []
for frame in source_frames:
    frame = frame.copy()
    frame.columns = [str(column).strip().lower().replace(' ', '_').replace('-', '_') for column in frame.columns]
    frame = frame.reindex(columns=EXPECTED_COLUMNS + ['_source'])
    aligned_frames.append(frame)
if not aligned_frames:
    raise FileNotFoundError('Tidak ada file pengamatan yang berhasil dibaca dari data/raw')
df_raw = pd.concat(aligned_frames, ignore_index=True, sort=False)
print('Gabungan seluruh pengamatan:', df_raw.shape)
display(pd.json_normalize(sensor_config).head())
display(df_raw.head())
```

## 2. Profil Awal Data

Profil dihitung setelah penggabungan dan sebelum pembersihan. Duplikat pengamatan dihitung dengan kunci sensor-waktu; jumlah missing ditampilkan per kolom.

```python
print('Jumlah baris dan kolom:', df_raw.shape)
print('Duplikat kunci sensor-waktu:', df_raw.duplicated(['sensor_id', 'timestamp']).sum())
profile_initial = pd.DataFrame({
    'dtype': df_raw.dtypes.astype(str),
    'missing': df_raw.isna().sum(),
    'missing_pct': (df_raw.isna().mean() * 100).round(2),
    'unique': df_raw.nunique(dropna=True),
})
display(profile_initial)
print('Total sel missing:', int(df_raw.isna().sum().sum()))
print('Baris per sumber:')
display(df_raw['_source'].value_counts(dropna=False).rename_axis('source').to_frame('rows'))
```

## 3. Cleaning dan Deteksi Outlier

Tipe waktu dan sensor numerik dikonversi secara eksplisit. Lokasi dinormalisasi ke Title Case, status ke huruf kecil. IQR memakai batas 1.5 kali rentang antarkuartil dan Z-score memakai ambang absolut 3. Kedua hasil dibandingkan per variabel. Outlier statistik hanya diberi flag; pelanggaran batas fisik dibuat missing.

```python
df_stage = df_raw.copy()
df_stage['timestamp'] = pd.to_datetime(df_stage['timestamp'], errors='coerce')
for column in MEASURE_COLUMNS:
    df_stage[column] = pd.to_numeric(df_stage[column], errors='coerce')
for column in ['sensor_id', 'location', 'status']:
    df_stage[column] = df_stage[column].astype('string').str.strip()
df_stage['sensor_id'] = df_stage['sensor_id'].str.upper()
df_stage['location'] = df_stage['location'].str.title()
df_stage['status'] = df_stage['status'].str.lower()

invalid_identity = df_stage['timestamp'].isna() | df_stage['sensor_id'].isna()
print('Baris tanpa timestamp/sensor_id yang dikeluarkan:', int(invalid_identity.sum()))
df_clean = df_stage.loc[~invalid_identity].copy()
duplicates_removed = int(df_clean.duplicated(['sensor_id', 'timestamp']).sum())
df_clean = df_clean.drop_duplicates(['sensor_id', 'timestamp'], keep='first')
df_clean = df_clean.sort_values(['sensor_id', 'timestamp']).reset_index(drop=True)
print('Duplikat pengamatan dihapus:', duplicates_removed)

outlier_rows = []
iqr_any = pd.Series(False, index=df_clean.index)
zscore_any = pd.Series(False, index=df_clean.index)
temperature_before_cleaning = df_clean['temperature_celsius'].copy()
for column in MEASURE_COLUMNS:
    values = df_clean[column]
    q1, q3 = values.quantile([0.25, 0.75])
    spread = q3 - q1
    iqr_mask = values.lt(q1 - 1.5 * spread) | values.gt(q3 + 1.5 * spread)
    standard_deviation = values.std(ddof=0)
    z_values = (values - values.mean()) / standard_deviation if standard_deviation else values * 0
    z_mask = z_values.abs().gt(3)
    iqr_any |= iqr_mask.fillna(False)
    zscore_any |= z_mask.fillna(False)
    outlier_rows.append({
        'variable': column,
        'IQR_count': int(iqr_mask.sum()),
        'Zscore_count': int(z_mask.sum()),
        'overlap_count': int((iqr_mask & z_mask).sum()),
        'union_count': int((iqr_mask | z_mask).sum()),
    })
outlier_comparison = pd.DataFrame(outlier_rows).set_index('variable')
df_clean['outlier_iqr_flag'] = iqr_any
df_clean['outlier_zscore_flag'] = zscore_any
display(outlier_comparison)
print('Baris dengan flag IQR:', int(iqr_any.sum()), '| Z-score:', int(zscore_any.sum()))

invalid_counts = {}
for column, (lower, upper) in VALID_RANGES.items():
    invalid = df_clean[column].notna() & ~df_clean[column].between(lower, upper)
    invalid_counts[column] = int(invalid.sum())
    df_clean.loc[invalid, column] = np.nan
print('Nilai di luar batas fisik yang diubah menjadi missing:')
display(pd.Series(invalid_counts, name='invalid_physical_values').to_frame())
print('Keputusan: outlier statistik dipertahankan dan ditandai; hanya pelanggaran batas fisik yang dikosongkan.')
```

## 4. Penanganan Missing Values

Dua strategi diterapkan dalam urutan waktu per sensor: forward fill untuk meneruskan pembacaan terakhir, lalu interpolasi linear untuk celah yang tersisa. Jika masih ada nilai kosong di awal/akhir atau sensor tidak memiliki pembacaan, imputasi median lokasi dan median keseluruhan menjadi fallback. Statistik deskriptif dibandingkan sebelum dan sesudah.

```python
stats_before_imputation = df_clean[MEASURE_COLUMNS].describe().T[['mean', 'std', 'min', '50%', 'max']]
missing_before = df_clean[MEASURE_COLUMNS].isna().sum()

df_clean[MEASURE_COLUMNS] = df_clean.groupby('sensor_id', observed=True)[MEASURE_COLUMNS].ffill()
for column in MEASURE_COLUMNS:
    df_clean[column] = df_clean.groupby('sensor_id', observed=True)[column].transform(
        lambda values: values.interpolate(method='linear', limit_direction='both')
    )
    location_median = df_clean.groupby('location', observed=True)[column].transform('median')
    df_clean[column] = df_clean[column].fillna(location_median).fillna(df_clean[column].median())

if df_clean['location'].isna().any():
    df_clean['location'] = df_clean['location'].fillna('Unknown')
if df_clean['status'].isna().any():
    status_mode = df_clean['status'].mode()
    df_clean['status'] = df_clean['status'].fillna(status_mode.iloc[0] if not status_mode.empty else 'unknown')
stats_after_imputation = df_clean[MEASURE_COLUMNS].describe().T[['mean', 'std', 'min', '50%', 'max']]
missing_comparison = pd.DataFrame({'missing_before': missing_before, 'missing_after': df_clean[MEASURE_COLUMNS].isna().sum()})
print('Jumlah nilai yang diisi per variabel:')
display(missing_comparison.assign(filled=missing_comparison['missing_before'] - missing_comparison['missing_after']))
print('Statistik sebelum imputasi:')
display(stats_before_imputation.round(3))
print('Statistik sesudah imputasi:')
display(stats_after_imputation.round(3))
print('Total missing pada empat pembacaan sensor:', int(df_clean[MEASURE_COLUMNS].isna().sum().sum()))
```

## 5. Feature Engineering

Fitur waktu mencakup jam, hari, minggu ISO, dan periode pagi/siang/sore/malam. Ditambahkan konversi suhu, kategori AQI, heat index, comfort index, dan rolling mean suhu 24 jam per sensor. Fitur orisinal `environmental_alert_score` memberi bobot risiko panas 25 poin, kelembapan tinggi 25 poin, dan AQI tinggi 50 poin; skor 0-100 memudahkan prioritas pemeriksaan lingkungan.

```python
df_final = df_clean.copy().sort_values(['sensor_id', 'timestamp']).reset_index(drop=True)
time_values = df_final['timestamp'].dt
df_final['hour'] = time_values.hour.astype('int8')
df_final['day'] = time_values.day.astype('int8')
df_final['week'] = time_values.isocalendar().week.astype('int16')
df_final['day_of_week'] = time_values.dayofweek.astype('int8')
df_final['time_period'] = np.select(
    [df_final['hour'].between(5, 10), df_final['hour'].between(11, 14), df_final['hour'].between(15, 17)],
    ['pagi', 'siang', 'sore'], default='malam'
)
df_final['temperature_fahrenheit'] = df_final['temperature_celsius'] * 9 / 5 + 32
df_final['temperature_kelvin'] = df_final['temperature_celsius'] + 273.15
aqI_bins = [-np.inf, 50, 100, 150, 200, 300, np.inf]
aqi_labels = ['Baik', 'Sedang', 'Tidak Sehat Sensitif', 'Tidak Sehat', 'Sangat Tidak Sehat', 'Berbahaya']
df_final['aqi_category'] = pd.cut(df_final['air_quality_aqi'], bins=aqI_bins, labels=aqi_labels)
temperature_f = df_final['temperature_fahrenheit']
humidity = df_final['humidity_percent']
heat_index_f = (-42.379 + 2.04901523 * temperature_f + 10.14333127 * humidity - 0.22475541 * temperature_f * humidity
                - 0.00683783 * temperature_f ** 2 - 0.05481717 * humidity ** 2 + 0.00122874 * temperature_f ** 2 * humidity
                + 0.00085282 * temperature_f * humidity ** 2 - 0.00000199 * temperature_f ** 2 * humidity ** 2)
df_final['heat_index_celsius'] = np.where(temperature_f >= 80, (heat_index_f - 32) * 5 / 9, df_final['temperature_celsius'])
temperature_penalty = (df_final['temperature_celsius'] - 22).abs() * 5
humidity_penalty = (df_final['humidity_percent'] - 50).abs() * 2
df_final['comfort_index'] = (100 - (temperature_penalty + humidity_penalty) / 2).clip(0, 100)
df_final['environmental_alert_score'] = (
    df_final['temperature_celsius'].gt(30).astype('int8') * 25
    + df_final['humidity_percent'].gt(80).astype('int8') * 25
    + df_final['air_quality_aqi'].gt(100).astype('int8') * 50
)
rolling_temperature = (
    df_final.groupby('sensor_id', observed=True).rolling('24h', on='timestamp', min_periods=1)['temperature_celsius']
    .mean().to_numpy()
)
df_final['temperature_rolling_24h_c'] = rolling_temperature
for column in ['sensor_id', 'location', 'status', 'time_period', 'aqi_category']:
    df_final[column] = df_final[column].astype('category')
print('Dimensi setelah feature engineering:', df_final.shape)
print('Fitur baru: hour, day, week, day_of_week, time_period, Fahrenheit, Kelvin, AQI, heat/comfort index, alert score, rolling 24h.')
display(df_final[['timestamp', 'sensor_id', 'hour', 'day', 'week', 'time_period', 'temperature_fahrenheit', 'temperature_kelvin', 'aqi_category', 'comfort_index', 'environmental_alert_score', 'temperature_rolling_24h_c']].head())
```

## 6. Validasi Kualitas Data dan Load

Completeness adalah persentase sel terisi pada skema sumber; uniqueness adalah proporsi kunci sensor-waktu yang unik; validity adalah pembacaan sensor dalam rentang fisik; consistency adalah persentase format waktu, ID sensor, lokasi, dan status yang konsisten. Skor sebelum ETL memakai data gabungan mentah dan sesudah ETL memakai data final.

```python
def quality_scores(frame):
    source_data = frame.reindex(columns=EXPECTED_COLUMNS)
    completeness = 100 * source_data.notna().to_numpy().mean() if source_data.size else 0
    uniqueness = 100 * (1 - source_data.duplicated(['sensor_id', 'timestamp']).mean()) if len(source_data) else 0
    valid_count = 0
    measured_count = 0
    for column, (lower, upper) in VALID_RANGES.items():
        values = pd.to_numeric(source_data[column], errors='coerce')
        present = values.notna()
        measured_count += int(present.sum())
        valid_count += int(values.between(lower, upper).sum())
    validity = 100 * valid_count / measured_count if measured_count else 100
    consistency_parts = []
    timestamp = pd.to_datetime(source_data['timestamp'], errors='coerce')
    timestamp_present = source_data['timestamp'].notna()
    consistency_parts.append(timestamp[timestamp_present].notna().mean() * 100 if timestamp_present.any() else 100)
    for column, normalizer in [('sensor_id', str.upper), ('location', str.title), ('status', str.lower)]:
        values = source_data[column].astype('string')
        present = values.notna()
        if present.any():
            normalized = values[present].str.strip().map(normalizer)
            consistency_parts.append(values[present].eq(normalized).mean() * 100)
        else:
            consistency_parts.append(100)
    consistency = float(np.mean(consistency_parts))
    return pd.Series({'completeness': completeness, 'uniqueness': uniqueness, 'validity': validity, 'consistency': consistency})

quality_comparison = pd.DataFrame({'sebelum_ETL': quality_scores(df_raw), 'sesudah_ETL': quality_scores(df_final)})
quality_comparison['perubahan_poin'] = quality_comparison['sesudah_ETL'] - quality_comparison['sebelum_ETL']
print('Skor kualitas data (persen):')
display(quality_comparison.round(2))

location_summary = df_final.groupby('location', observed=True).agg(
    reading_count=('sensor_id', 'size'), temperature_mean_c=('temperature_celsius', 'mean'),
    humidity_mean_pct=('humidity_percent', 'mean'), pressure_mean_hpa=('pressure_hpa', 'mean'),
    aqi_mean=('air_quality_aqi', 'mean'), comfort_mean=('comfort_index', 'mean')
).round(2).reset_index()
sensor_summary = df_final.groupby('sensor_id', observed=True).agg(
    reading_count=('timestamp', 'size'), location=('location', 'first'),
    temperature_mean_c=('temperature_celsius', 'mean'), temperature_std_c=('temperature_celsius', 'std'),
    humidity_mean_pct=('humidity_percent', 'mean'), aqi_mean=('air_quality_aqi', 'mean')
).round(2).reset_index()

csv_path = OUTPUT_DIR / 'Tugas2_sensor_bersih.csv'
excel_path = OUTPUT_DIR / 'Tugas2_ringkasan_sensor.xlsx'
json_path = OUTPUT_DIR / 'Tugas2_sensor_bersih.json'
parquet_path = OUTPUT_DIR / 'Tugas2_sensor_bersih.parquet'
df_final.to_csv(csv_path, index=False)
with pd.ExcelWriter(excel_path, engine='openpyxl') as writer:
    df_final.to_excel(writer, sheet_name='Data_Bersih', index=False)
    location_summary.to_excel(writer, sheet_name='Ringkasan_Lokasi', index=False)
    sensor_summary.to_excel(writer, sheet_name='Ringkasan_Sensor', index=False)
    quality_comparison.to_excel(writer, sheet_name='Kualitas_Data')
    outlier_comparison.to_excel(writer, sheet_name='Perbandingan_Outlier')
df_final.to_json(json_path, orient='records', date_format='iso', indent=2)
df_final.to_parquet(parquet_path, index=False)

output_files = [csv_path, excel_path, json_path, parquet_path]
file_sizes = pd.DataFrame({'file': [path.name for path in output_files], 'size_bytes': [path.stat().st_size for path in output_files]})
file_sizes['size_MB'] = (file_sizes['size_bytes'] / 1024 ** 2).round(3)
print('Bukti load dan ukuran berkas:')
display(file_sizes)
print('Sheet Excel:', pd.ExcelFile(excel_path).sheet_names)
print('Contoh CSV:')
display(pd.read_csv(csv_path, nrows=3))
print('Contoh JSON:')
display(pd.read_json(json_path).head(3))
print('Contoh Parquet:')
display(pd.read_parquet(parquet_path).head(3))
print('Ringkasan per lokasi:')
display(location_summary)
print('Ringkasan per sensor:')
display(sensor_summary.head(10))
```

## 7. Visualisasi dan Interpretasi

Visualisasi berikut menunjukkan tren suhu harian per sensor, korelasi antarvariabel numerik, serta distribusi suhu sebelum dan sesudah pembersihan/imputasi. Interpretasi berbasis ringkasan numerik dicetak setelah grafik.

```python
daily_temperature = df_final.assign(date=df_final['timestamp'].dt.floor('d')).groupby(
    ['date', 'sensor_id'], observed=True
)[ 'temperature_celsius' ].mean().reset_index()
fig, ax = plt.subplots(figsize=(12, 5))
sns.lineplot(data=daily_temperature, x='date', y='temperature_celsius', hue='sensor_id', linewidth=1, ax=ax)
ax.set(title='Tren Suhu Harian per Sensor', xlabel='Tanggal', ylabel='Suhu rata-rata (C)')
ax.legend(title='Sensor', bbox_to_anchor=(1.02, 1), loc='upper left')
fig.tight_layout()
plt.show()

correlation_columns = MEASURE_COLUMNS
fig, ax = plt.subplots(figsize=(7, 5))
sns.heatmap(df_final[correlation_columns].corr(), annot=True, cmap='vlag', center=0, fmt='.2f', ax=ax)
ax.set_title('Heatmap Korelasi Pembacaan Sensor')
fig.tight_layout()
plt.show()

temperature_plot = pd.concat([
    pd.DataFrame({'tahap': 'Sebelum cleaning', 'temperature_celsius': pd.to_numeric(df_raw['temperature_celsius'], errors='coerce')}),
    pd.DataFrame({'tahap': 'Sesudah ETL', 'temperature_celsius': df_final['temperature_celsius']})
]).dropna()
fig, ax = plt.subplots(figsize=(7, 5))
sns.boxplot(data=temperature_plot, x='tahap', y='temperature_celsius', showfliers=False, ax=ax)
ax.set(title='Distribusi Suhu Sebelum dan Sesudah ETL', xlabel='', ylabel='Suhu (C)')
fig.tight_layout()
plt.show()

daily_spread = daily_temperature.groupby('date')['temperature_celsius'].mean()
strongest_pair = df_final[correlation_columns].corr().abs().where(np.triu(np.ones((len(correlation_columns), len(correlation_columns))), k=1).astype(bool)).stack().idxmax()
print('Interpretasi tren: rerata harian agregat berkisar {:.2f} sampai {:.2f} C.'.format(daily_spread.min(), daily_spread.max()))
print('Interpretasi korelasi: pasangan dengan korelasi absolut terbesar adalah {} dan {}.'.format(*strongest_pair))
print('Interpretasi distribusi: grafik membandingkan data gabungan mentah dengan hasil final; outlier statistik tetap dipertahankan, sedangkan nilai di luar batas fisik telah diimputasi.')
print('Jumlah baris bertanda IQR/Z-score:', int(df_final['outlier_iqr_flag'].sum()), '/', int(df_final['outlier_zscore_flag'].sum()))
```
